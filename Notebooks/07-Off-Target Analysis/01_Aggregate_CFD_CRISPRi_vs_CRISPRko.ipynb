```python
#load crisprko data
crisprko_df = pd.read_csv('../../Data/externalData/RS3Val/deweirdt2022tiling_nonessentials_zscored.csv')
crisprko_df
```

```python
#merge with designs to get RS3 Sequence scores and context sequences
crisprko_designs = pd.read_table('../../Data/externalData/RS3Val/rs3val_nonessentialscleaned_scoredrs3seq_notarget-sgrna-designs.txt')
crisprko_scored = crisprko_df.merge(crisprko_designs[['sgRNA Sequence','sgRNA Context Sequence','On-Target Efficacy Score']].drop_duplicates(),
                                    on = 'sgRNA Sequence', how = 'inner')
crisprko_scored
```

```python
#since RS3 Validation tiling includes only guides with RS3 Sequence > 0.2, subset crispri data equivalently for fair comparison
crispri_active = crispri_df[(crispri_df['On-Target Efficacy Score'] > 0.2) | (crispri_df['Category'] == 'ONE_INTERGENICO_CTL')].copy()
crispri_active
```

```python
crispri_active['sgRNA Context Sequence'].dropna().to_csv('../../Data/offTargetAnalysis/crispri_noness_tiling_sequences.txt', sep = '\n', index=False, header=None)
crisprko_active['sgRNA Context Sequence'].dropna().to_csv('../../Data/externalData/RS3Val/crisprko_noness_tiling_sequences.txt', sep = '\n', index=False, header=None)
```

```python
#read in designs with aggregate cfd scores
crispri_agg_cfd_df = pd.read_table('../../Data/CRISPickRuns/crispri_guides_crisprko_designs.txt')
crisprko_agg_cfd_df = pd.read_table('../../Data/CRISPickRuns/crisprko_guides_crisprko_designs.txt')
crisprko_agg_cfd_df
```

```python
crispri_df = crispri_active.merge(crispri_agg_cfd_df[['sgRNA Sequence','Aggregate CFD Score']].drop_duplicates(),
                                  on = 'sgRNA Sequence', how = 'inner')
crisprko_df = crisprko_active.merge(crisprko_agg_cfd_df[['sgRNA Sequence','Aggregate CFD Score']].drop_duplicates(),
                                  on = 'sgRNA Sequence', how = 'inner')
crisprko_df['Aggregate CFD Score'].min()
```

Since a raw DNA sequence was provided as input, CRISPick counts the location of each guide in the input sequence as an off-target site, automatically increasing Aggregate CFD by 1. To account for that, we must subtract 1 from every Aggregate CFD score.

```python
crispri_df['Aggregate CFD Score - 1'] = crispri_df['Aggregate CFD Score'] - 1
crisprko_df['Aggregate CFD Score - 1'] = crisprko_df['Aggregate CFD Score'] - 1
```

```python
#add column to specify which df is which
crispri_df['Mechanism'] = 'CRISPRi'
crisprko_df['Mechanism'] = 'CRISPRko'

#combine
combined_agg_cfd = pd.concat([crispri_df[['Aggregate CFD Score - 1','Mechanism','promiscuous']],
                              crisprko_df[['Aggregate CFD Score - 1','Mechanism','promiscuous']]]).drop_duplicates().reset_index(drop=True)
combined_agg_cfd
```

```python
thresholds = np.arange(0, 20, 0.1)
group_df_list = []
for group, df in combined_agg_cfd.groupby('Mechanism'):
    f1_scores = []
    for threshold in thresholds:
        df['pred_promiscuous'] = df['Aggregate CFD Score - 1'] > threshold
        f1_scores.append(f1_score(df['promiscuous'], df['pred_promiscuous'], average='macro'))
    group_df = pd.DataFrame({'Threshold': thresholds, 'F1 Score': f1_scores})
    group_df['Mechanism'] = group
    group_df_list.append(group_df)

plot_df = pd.concat(group_df_list)
color_map = {'CRISPRi':'red','CRISPRko':'blue'}
sns.lineplot(data = plot_df, x = 'Threshold', y = 'F1 Score', hue = 'Mechanism', palette = color_map)
plt.hlines(y = 0.5, xmin = min(thresholds), xmax = max(thresholds), color = 'black', linestyles = 'dotted')
sns.despine()

plt.xlabel('Aggregate CFD Threshold',fontsize=12)
plt.ylabel('Macro-averaged F1 Score',fontsize=12)
plt.title('Classification of promiscuous nonessential guides',fontsize=14)
plt.legend(title='Mechanism', bbox_to_anchor = [1,1], title_fontsize=12, fontsize=10)

gpp.savefig('../../Figures/Figure 5/agg_cfd_crispri_vs_crisprko_f1_score.pdf', bbox_inches = 'tight', dpi = 300)
gpp.savefig('../../Figures/Figure 5/agg_cfd_crispri_vs_crisprko_f1_score.png', bbox_inches = 'tight', dpi = 600)
```

```python
max_f1_df = plot_df.groupby('Mechanism')['F1 Score'].max().reset_index()
sns.barplot(data = max_f1_df, x = 'Mechanism', y = 'F1 Score', color = 'blue', order = ['CRISPRko','CRISPRi'])
plt.hlines(y = 0.5, xmin = -1, xmax = 2, color = 'black', linestyles = '--', label = 'Random performance')
plt.legend(fontsize=12)
plt.ylabel('Maximum Macro-averaged F1 Score',fontsize=16)
plt.xlabel('Mechanism',fontsize=16)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.title('CRISPick Aggregate CFD performance in classifying promiscuous nonessential-targeting guides',fontsize=14)
gpp.savefig('../../Figures/Figure 5/agg_cfd_crispri_vs_crisprko_max_f1_score.png', bbox_inches = 'tight', dpi = 600)
```

```python
#what about cfd100 off-target sites? get this info from the original design file 
crispick = pd.read_csv('../../Data/primaryLibrary/essentials_nonessentials_CRISPRi_tiling_designs_v2.csv')
crispick.columns
```

```python
#calculate total number of cfd100 off-target sites per guide
cfd100_cols = crispick.columns[crispick.columns.str.endswith('Bin I Matches')].tolist()
crispick['cfd100_ots'] = crispick[cfd100_cols].sum(axis=1)
crispri_cfd100_ots = crispick[['sgRNA Sequence','cfd100_ots']].drop_duplicates()
crispri_cfd100_ots
```

```python
crispri_with_cfd100_ots = crispri_df.merge(crispri_cfd100_ots, on = 'sgRNA Sequence', how = 'inner')
crispri_with_cfd100_ots
```

```python
#at various CFD100 off-target site cutoffs, how many promiscuous guides are captured? How significant is the overrepresentation of promiscuity?

total_promiscuous = crispri_with_cfd100_ots['promiscuous'].sum()
total_nonpromiscuous = (~crispri_with_cfd100_ots['promiscuous']).sum()

cutoffs = list(np.arange(0,16,1))
pvalues = []
for cutoff in cutoffs:
    group_df = crispri_with_cfd100_ots[crispri_with_cfd100_ots['cfd100_ots'] > cutoff]
    crispri_with_cfd100_ots['>cutoff CFD100 matches'] = crispri_with_cfd100_ots['cfd100_ots'] > cutoff
    num_promiscuous = group_df['promiscuous'].sum()
    num_nonpromiscuous = (~group_df['promiscuous']).sum()
    pvalue = fisher_exact(np.array([[num_promiscuous,num_nonpromiscuous],[total_promiscuous,total_nonpromiscuous]]), alternative = 'greater').pvalue
    pvalues.append(pvalue)

plt.scatter(cutoffs, pvalues)
plt.plot(cutoffs, pvalues)

sns.despine()
plt.xlabel('Number of CFD100 Matches Cutoff')
plt.ylabel('Fisher Exact test p-value')
plt.title('Testing association between having >cutoff CFD100 matches and guide promiscuity')

gpp.savefig('../../Figures/Figure 5/cfd100_matches_fisher_exact_pvalues.png', bbox_inches = 'tight', dpi = 300)
gpp.savefig('../../Figures/Figure 5/cfd100_matches_fisher_exact_pvalues.pdf', bbox_inches = 'tight', dpi = 300)
```

```python
def count_overlapping_matches(seq,pattern):
    count = 0
    length = len(pattern)
    for i in range(len(seq)-1):
        if seq[i:(i+length)] == pattern:
            count +=1
    return count
```

```python
crispri_with_cfd100_ots['seed_score'] = crispri_with_cfd100_ots['sgRNA Sequence'].str[-12:].apply(lambda seq: count_overlapping_matches(seq,'GG'))
crispri_with_cfd100_ots_no_GGs = crispri_with_cfd100_ots[crispri_with_cfd100_ots['seed_score'] == 0].copy()
crispri_with_cfd100_ots_no_GGs
```

```python
#at various CFD100 off-target site cutoffs, how many promiscuous guides are captured? How significant is the overrepresentation of promiscuity?

total_promiscuous = crispri_with_cfd100_ots_no_GGs['promiscuous'].sum()
total_nonpromiscuous = (~crispri_with_cfd100_ots_no_GGs['promiscuous']).sum()

cutoffs = list(np.arange(0,16,1))
pvalues = []
for cutoff in cutoffs:
    group_df = crispri_with_cfd100_ots[crispri_with_cfd100_ots['cfd100_ots'] > cutoff]
    crispri_with_cfd100_ots['>cutoff CFD100 matches'] = crispri_with_cfd100_ots['cfd100_ots'] > cutoff
    num_promiscuous = group_df['promiscuous'].sum()
    num_nonpromiscuous = (~group_df['promiscuous']).sum()
    if num_promiscuous < 10:
        print('Cutoff of ' + str(cutoff) + ' yields ' + str(num_promiscuous) + ' promiscuous guides')
    pvalue = fisher_exact(np.array([[num_promiscuous,num_nonpromiscuous],[total_promiscuous,total_nonpromiscuous]]), alternative = 'greater').pvalue
    pvalues.append(pvalue)

plt.scatter(cutoffs, pvalues)
plt.plot(cutoffs, pvalues)

sns.despine()
plt.xlabel('Number of CFD100 Matches Cutoff')
plt.ylabel('Fisher Exact test p-value')
plt.title('Testing association between having >cutoff CFD100 matches and guide promiscuity\n(only guides with 0 GGs)')

gpp.savefig('../../Figures/Figure 5/cfd100_matches_no_GGs_fisher_exact_pvalues.png', bbox_inches = 'tight', dpi = 300)
gpp.savefig('../../Figures/Figure 5/cfd100_matches_no_GGs_fisher_exact_pvalues.pdf', bbox_inches = 'tight', dpi = 300)
```
